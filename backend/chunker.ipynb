```python
notes = """
Normalization is the process of organizing a database to reduce duplicate data.
It splits big tables into smaller ones and links them using keys.

A join combines rows from two tables based on a related column.
An inner join returns only the rows that match in both tables.
A left join returns all rows from the left table, even without a match.

A deadlock happens when two processes wait for each other forever.
Process A holds resource 1 and needs resource 2.
Process B holds resource 2 and needs resource 1.
Neither can continue, so both are stuck.
"""

print(len(notes), "characters")
```

```python
def chunk_fixed(text, size=150):
    """Cut text every `size` characters."""
    return [text[i:i + size] for i in range(0, len(text), size)]


for i, chunk in enumerate(chunk_fixed(notes)):
    print(f"--- chunk {i} ---")
    print(chunk)
```

```python
import re


def chunk_paragraphs(text, max_chars=500):
    """Split text into paragraph chunks, each at most `max_chars` long
    (unless a single sentence is longer than that)."""
    chunks = []

    for para in re.split(r"\n\s*\n", text):   # split at blank lines
        para = para.strip()
        if not para:                          # skip empty pieces
            continue

        if len(para) <= max_chars:            # short enough: keep whole
            chunks.append(para)
            continue

        # too long: pack sentences into chunks up to max_chars
        current = ""
        for sentence in re.split(r"(?<=[.!?])\s+", para):
            if current and len(current) + 1 + len(sentence) > max_chars:
                chunks.append(current)
                current = sentence
            else:
                current = f"{current} {sentence}".strip()
        if current:
            chunks.append(current)

    return chunks


for i, chunk in enumerate(chunk_paragraphs(notes)):
    print(f"--- chunk {i} ---")
    print(chunk)
```

```python
def test_chunk_paragraphs():
    # empty input gives no chunks
    assert chunk_paragraphs("") == []
    assert chunk_paragraphs("   \n\n  ") == []

    # two paragraphs give two chunks
    assert chunk_paragraphs("First idea.\n\nSecond idea.") == ["First idea.", "Second idea."]

    # extra blank lines are ignored
    assert chunk_paragraphs("A.\n\n\n\n\nB.") == ["A.", "B."]

    # a long paragraph is split, and no chunk is too big
    long_para = "This is a sentence. " * 50
    result = chunk_paragraphs(long_para, max_chars=100)
    assert len(result) > 1
    assert all(len(c) <= 100 for c in result)

    # the sample notes give 3 chunks
    assert len(chunk_paragraphs(notes)) == 3

    print("All tests passed")


test_chunk_paragraphs()
```

```python
my_text = """Normalization is the process of organizing a database to reduce duplicate data.
It splits big tables into smaller ones and links them using keys.

A join combines rows from two tables based on a related column.
An inner join returns only the rows that match in both tables.
A left join returns all rows from the left table, even without a match.

A deadlock happens when two processes wait for each other forever.
Process A holds resource 1 and needs resource 2.
Process B holds resource 2 and needs resource 1.
Neither can continue, so both are stuck.

"""

my_chunks = chunk_paragraphs(my_text)
print(len(my_chunks), "chunks")
for i, c in enumerate(my_chunks):
    print(f"--- chunk {i} ({len(c)} chars) ---")
    print(c)
```
